# OpenEarthMap multi-GPU training on Kaggle
Enable **2 GPUs** in Kaggle before running this notebook. The launcher reuses the project's real `run_training` function; it does not duplicate the training loop. Start from a fresh kernel and do not initialize CUDA before the launcher cell.


```python
PROJECT_ROOT = '/kaggle/working/OEM_Segmentation'  # change if your repo is mounted elsewhere
%cd $PROJECT_ROOT
!python -m pip install --no-build-isolation -r requirements.txt
```

```python
from accelerate import notebook_launcher
from oemseg.config import parse_args
from oemseg.engine.trainer import run_training

DATA_ROOT = '/kaggle/input/openearthmap/OpenEarthMap'  # point this at train.txt/val.txt + region folders
OUTPUT_ROOT = '/kaggle/working/outputs'
TRAIN_ARGS = [
    '--data-root', DATA_ROOT,
    '--output-root', OUTPUT_ROOT,
    '--model', 'unetpp',
    '--model-variant', 'resnet18',
    '--val-fraction', '0.1',
    '--mixed-precision', 'fp16',
]

def training_function(train_args):
    return run_training(parse_args(train_args))

notebook_launcher(
    training_function,
    args=(TRAIN_ARGS,),
    num_processes=2,
    mixed_precision='fp16',
)
```
